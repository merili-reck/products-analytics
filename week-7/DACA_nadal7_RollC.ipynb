## Roll B – Andmete puhastamine

Puhastamise käigus:

- eemaldasin 988 rida, kus puudus `customer_id`;
- teisendasin `sale_date` kuupäevaformaati;

Pärast puhastamist:
- ridu: 8950

- unikaalseid kliente: 2540
- kuupäevavahemik: 01.01.2023–03.12.2026

Puhastatud DataFrame on valmis RFM-analüüsiks.

```python
import pandas as pd
```

```python
today = pd.to_datetime('2026-12-04')
today
```

```python
df = df.dropna(subset=['customer_id', 'sale_date', 'total_price'])
```

```python
df['sale_date'] = pd.to_datetime(df['sale_date'], format='mixed', dayfirst=True)
```

```python
(df['total_price'] < 0).sum()
```

```python
df = df[df['total_price'] > 0]
```

```python
recency = df.groupby('customer_id')['sale_date'].max().reset_index()
recency.columns = ['customer_id', 'last_purchase']

recency['recency_days'] = (today - recency['last_purchase']).dt.days

recency.head()
```

```python
frequency = df.groupby('customer_id').size().reset_index(name='frequency')

frequency.head()
```

```python
monetary = df.groupby('customer_id')['total_price'].sum().reset_index()
monetary.columns = ['customer_id', 'monetary']

monetary.head()
```

```python
rfm = recency[['customer_id', 'recency_days']].merge(
    frequency, on='customer_id'
).merge(
    monetary, on='customer_id'
)

rfm.head()
```

```python
rfm['R_score'] = pd.qcut(
    rfm['recency_days'],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm['F_score'] = pd.qcut(
    rfm['frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm['M_score'] = pd.qcut(
    rfm['monetary'],
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm.head()
```

```python
rfm['R_score'] = rfm['R_score'].astype(int)
rfm['F_score'] = rfm['F_score'].astype(int)
rfm['M_score'] = rfm['M_score'].astype(int)

rfm['RFM_Score'] = (
    rfm['R_score'] +
    rfm['F_score'] +
    rfm['M_score']
)

rfm.head()
```

```python
def get_segment(score):
    if score >= 13:
        return 'VIP Champions'
    elif score >= 10:
        return 'Loyal Customers'
    elif score >= 7:
        return 'Potential Loyalists'
    elif score >= 4:
        return 'At Risk'
    else:
        return 'Lost'

rfm['Segment'] = rfm['RFM_Score'].apply(get_segment)

rfm.head()
```

```python
segment_counts = rfm['Segment'].value_counts().reset_index()
segment_counts.columns = ['Segment', 'Klientide arv']

segment_counts['Osakaal %'] = (
    segment_counts['Klientide arv'] / segment_counts['Klientide arv'].sum() * 100
).round(2)

segment_counts
```
