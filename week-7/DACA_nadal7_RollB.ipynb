```python
import pandas as pd
df_sales = pd.read_csv('csv/sales.csv')
print(df_sales.shape)
print(df_sales.head())
df_customers = pd.read_csv('csv/customers.csv')
print(df_customers.shape)
print(df_customers.head())
df = pd.merge(df_sales, df_customers, on='customer_id', how='left')
print(df.shape)
print(df.dtypes)
print(df.head())
```
