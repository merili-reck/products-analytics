## Roll B – Andmete puhastamine

Puhastamise käigus:
- eemaldasin 5116 duplikaati `invoice_id` järgi;
- eemaldasin 988 rida, kus puudus `customer_id`;
- teisendasin `sale_date` kuupäevaformaati;
- eemaldasin 180 negatiivse `total_price` väärtusega rida.

Pärast puhastamist:
- ridu: 8950
- veerge: 19
- unikaalseid kliente: 2540
- kuupäevavahemik: 01.01.2023–03.12.2026

Puhastatud DataFrame on valmis RFM-analüüsiks.

```python
import pandas as pd
```

```python
today = pd.to_datetime('2026-12-04')
today
```

```python
df = df.dropna(subset=['customer_id', 'sale_date', 'total_price'])
```

```python
df['sale_date'] = pd.to_datetime(df['sale_date'], format='mixed', dayfirst=True)
```

```python
(df['total_price'] < 0).sum()
```

```python
df = df[df['total_price'] > 0]
```

```python
recency = df.groupby('customer_id')['sale_date'].max().reset_index()
recency.columns = ['customer_id', 'last_purchase']

recency['recency_days'] = (today - recency['last_purchase']).dt.days

recency.head()
```

```python
frequency = df.groupby('customer_id').size().reset_index(name='frequency')

frequency.head()
```

```python
monetary = df.groupby('customer_id')['total_price'].sum().reset_index()
monetary.columns = ['customer_id', 'monetary']

monetary.head()
```

```python
rfm = recency[['customer_id', 'recency_days']].merge(
    frequency, on='customer_id'
).merge(
    monetary, on='customer_id'
)

rfm.head()
```

```python
rfm['R_score'] = pd.qcut(
    rfm['recency_days'],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm['F_score'] = pd.qcut(
    rfm['frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm['M_score'] = pd.qcut(
    rfm['monetary'],
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm.head()
```

```python
rfm['R_score'] = rfm['R_score'].astype(int)
rfm['F_score'] = rfm['F_score'].astype(int)
rfm['M_score'] = rfm['M_score'].astype(int)

rfm['RFM_Score'] = (
    rfm['R_score'] +
    rfm['F_score'] +
    rfm['M_score']
)

rfm.head()
```

```python
def get_segment(score):
    if score >= 13:
        return 'VIP Champions'
    elif score >= 10:
        return 'Loyal Customers'
    elif score >= 7:
        return 'Potential Loyalists'
    elif score >= 4:
        return 'At Risk'
    else:
        return 'Lost'

rfm['Segment'] = rfm['RFM_Score'].apply(get_segment)

rfm.head()
```

```python
segment_counts = rfm['Segment'].value_counts().reset_index()
segment_counts.columns = ['Segment', 'Klientide arv']

segment_counts['Osakaal %'] = (
    segment_counts['Klientide arv'] / segment_counts['Klientide arv'].sum() * 100
).round(2)

segment_counts
```

```python
# Diagramm 1: Segmentide jaotus
fig1 = px.bar(
    segment_counts,
    x='Segment',
    y='Klientide arv',
    color='Segment',
    title='UrbanStyle Kliendisegmentide jaotus',
    labels={'Segment': 'Segment', 'Klientide arv': 'Klientide arv'},
    text='Klientide arv'
)
fig1.update_layout(showlegend=False)
fig1.show()

# Diagramm 2: Hajuvusdiagramm
fig2 = px.scatter(
    rfm,
    x='recency_days',
    y='monetary',
    color='Segment',
    size='frequency',
    title='UrbanStyle Kliendisegmendid (RFM)',
    labels={
        'recency_days': 'Päevi viimasest ostust',
        'monetary': 'Kogukulutus (EUR)',
        'frequency': 'Ostude arv',
        'Segment': 'Segment'
    }
)
fig2.show()

# Diagramm 3: Top 10 VIP klienti
top_vip = rfm[rfm['Segment'] == 'VIP Champions'].nlargest(10, 'monetary')
fig3 = px.bar(
    top_vip,
    x='customer_id',
    y='monetary',
    title='Top 10 VIP klienti kogukulutuse järgi',
    labels={'customer_id': 'Kliendi ID', 'monetary': 'Kogukulutus (EUR)'},
    text='monetary'
)
fig3.update_xaxes(type='category')
fig3.show()
```

```python
# Äritõlgendus Markole: VIP ja At-Risk ülevaade
vip_count = (rfm['Segment'] == 'VIP Champions').sum()
vip_share = (rfm.loc[rfm['Segment'] == 'VIP Champions', 'monetary'].sum() / rfm['monetary'].sum() * 100).round(1)
atrisk_count = (rfm['Segment'] == 'At Risk').sum()
atrisk_share = (rfm.loc[rfm['Segment'] == 'At Risk', 'monetary'].sum() / rfm['monetary'].sum() * 100).round(1)

print(f"VIP Champions: {vip_count} klienti")
print(f"VIP käibeosakaal: {vip_share}%")
print(f"At-Risk: {atrisk_count} klienti")
print(f"At-Risk käibeosakaal: {atrisk_share}%")
```
